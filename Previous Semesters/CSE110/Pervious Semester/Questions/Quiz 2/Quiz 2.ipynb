**Quiz 2 Set A**

```python
input_str = input()
upper = ''
lower = ''
for a in input_str:
  if ord(a)>=65 and ord(a)<90:
    upper+=a
  elif ord(a)>=97 and ord(a)<=122:
    lower+=a
print('Uppercase:', upper)
print('Lowercase:', lower)
print('New String:', upper+lower)
```

**Quiz 2 Set B**

```python
inp_str = input()
odd_str = ''
count = 0
for a in inp_str:
  if ord(a)%2 !=0:
    odd_str+=a
    count+=1
print('Odd characters:', odd_str)
print('Number of odd ASCII value characters:', count )
```
