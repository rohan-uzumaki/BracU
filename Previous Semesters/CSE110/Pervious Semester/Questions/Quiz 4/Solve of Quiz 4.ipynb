```python
given_list = [1,3,2,2,3,7,6,4]

def index_number(l,a):
  index = None
  for i in range(len(l)):
    if l[i] == a:
      index = i
      break
  return index

number = int(input())
val = index_number(given_list,number)
if val == None:
  print('Number of found!')
else:
  print('Index =', val)
```

```python
def sqare_shape_generator(val):
  if val%2 == 0:
      for a in range(val):
        for b in range(val):
           print('*', end = '')
        
  else:
       for a in range(val):
        for b in range(val):
           print('#', end = '')
        print()   

number = int(input())
sqare_shape_generator(number)
```

```python
def area_calculator(h, w):
  if h == w:
    area = float(h*w)
  else:
    area = 0.5 * h * w
  return area
he = int(input())
wi = int(input())
val = area_calculator(he,wi)
print(val)
```

```python
def palindrome_checker(value):
  new_val = ''
  for a in value:
    if a != ' ':
      new_val +=a
  print(new_val)
  new = ''
  for a in range(len(new_val)-1,-1,-1):
       new+=new_val[a]
  if new_val == new:
    return 'Palindrome'
  else:
    return 'Not a palindrome'

my_string = input()
print(palindrome_checker(my_string))
```
