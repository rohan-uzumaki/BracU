```python
n = int(input())
sum = 0
for a in range(2):
  digit = n%10
  n = n//10
  sum+=digit
print(sum)
if sum%4 == 0:
  print('Happy Number')
else:
  print('Not a happy number')
```

**Quiz 1 Set 3 Question 1**

```python
n = int(input())
sum = 0
counter = 0
print('Divisors: ', end = '')
for a in range(2,n +1):
  if n%a == 0:
    sum+=a
    counter+=1
    print(a, end = ' ')
print()
if sum%2 != 0:
  print(n/counter)
else:
  print(n)
```

**Quiz 1 Set 4 Question 1**

```python
N = int(input())
sum = 0
for a in range(3, N+1):
  if a%2 == 0:
    if N%a == 0:
      print(a, end = ' ')
      sum+=a
print()
if sum% 5 == 0:
  print(sum/5)
else:
  print(N)
```

**Quiz 1 Set 5 Question 1**

```python
items = int(input())
total_price = 0
discounted_price = 0
for a in range(items):
  price = int(input())
  total_price +=price

if total_price % 5 == 0:
  discounted_price = total_price - total_price * 0.05
else:
  discounted_price = total_price - total_price * 0.10

print('Price before discount:', total_price)
print('Price after discount:', discounted_price)

```

```python
items = int(input())
div_sum = 0
total_cost = 0
for a in range(items):
  price = int(input())
  if price%2 == 0:
    div_sum+=price
  total_cost+=price

discounted_price = total_cost - (div_sum*0.05)
print('Price before discount:', total_cost)
print('Price after discount:', discounted_price)
```
